```python
import pandas as pd
from  sklearn.datasets import load_iris
iris=load_iris()
```

```python
dir(iris)
```

```python
iris.feature_names
```

```python
iris.target_names
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df.head()
```

```python
df['target']=iris.target
df.head()
```

```python
df[df.target==1].head()
```

```python
df[df.target==2].head()
```

```python
df['flower_name']=df.target.apply(lambda x:iris.target_names[x])
df.head()
```

```python
df[45:55]
```

```python
df0=df[0:50]
df1=df[50:100]
df2=df[100:]
```

```python
from matplotlib import pyplot as plt
```

Sepal length vs Sepal Width (Setosa vs Versicolor)

```python

plt.xlabel('sepal length (cm)')
plt.ylabel('sepal width (cm)')
plt.scatter(df0['sepal length (cm)'],df0['sepal width (cm)'],marker="+",color='red')
plt.scatter(df1['sepal length (cm)'],df1['sepal width (cm)'],marker=".",color='green')


```

Petal length vs Pepal Width (Setosa vs Versicolor)

```python
plt.xlabel('petal length (cm)')
plt.ylabel('petal width (cm)')
plt.scatter(df0['petal length (cm)'],df0['petal width (cm)'],marker='d',color='blue')
plt.scatter(df1['petal length (cm)'],df1['petal width (cm)'],marker='*',color='purple')
```

```python
x=df.drop(['target','flower_name'],axis='columns')
y=df.target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
len(X_test)
```

```python
len(X_train)
```

```python
from sklearn.svm import SVC
model=SVC()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict([[4.8,3.0,1.5,0.3]])
```

Tune parameters

1. Regularization (C)

```python
model_C=SVC(C=100)
model_C.fit(X_train,y_train)
model_C.score(X_test,y_test)
```

2. Gamma

```python
model_g=SVC(gamma=100)
model_g.fit(X_train,y_train)
model_g.score(X_test,y_test)
```

3. Kernel

```python
model_linear_kernel=SVC(kernel='linear')
model_linear_kernel.fit(X_train,y_train)
model_linear_kernel.score(X_test,y_test)
```
